```python
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
import matplotlib.pyplot as plt
import numpy as np
import math
import os
import glob
```

```python
img_A = cv.imread('/content/drive/My Drive/PCVK 2026/P5/citra_A_10x10.png', cv.IMREAD_GRAYSCALE)
img_B = cv.imread('/content/drive/My Drive/PCVK 2026/P5/citra_B_10x10.png', cv.IMREAD_GRAYSCALE)

hist_A = np.zeros(256, dtype=int)
hist_B = np.zeros(256, dtype=int)

for y in range(img_A.shape[0]):
    for x in range(img_A.shape[1]):
        pixel = img_A[y, x]
        hist_A[pixel] += 1

for y in range(img_B.shape[0]):
    for x in range(img_B.shape[1]):
        pixel = img_B[y, x]
        hist_B[pixel] += 1

plt.figure(figsize=(12, 8))

# Citra A
plt.subplot(2, 2, 1)
plt.imshow(img_A, cmap='gray')
plt.title('Citra A')
plt.axis('off')

# Histogram A
plt.subplot(2, 2, 2)
plt.bar(range(256), hist_A)
plt.xlabel('Intensitas Piksel')
plt.ylabel('Jumlah Piksel')
plt.title('Histogram Citra A')

# Citra B
plt.subplot(2, 2, 3)
plt.imshow(img_B, cmap='gray')
plt.title('Citra B')
plt.axis('off')

# Histogram B
plt.subplot(2, 2, 4)
plt.bar(range(256), hist_B)
plt.xlabel('Intensitas Piksel')
plt.ylabel('Jumlah Piksel')
plt.title('Histogram Citra B')

plt.tight_layout()
plt.show()
```

```python
#Praktikum E-1
img = cv.imread('/content/drive/My Drive/PCVK 2026/P5/lena.jpg')
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)

height, width, depth = np.shape(img)
names = np.arange(256)

red = [0]*256
green = [0]*256
blue = [0]*256

for y in range(0,height) :
  for x in range(0,width) :
    red[img[y][x][0]] += 1
    green[img[y][x][1]] += 1
    blue[img[y][x][2]] += 1

names = np.arange(256)
fig, axs = plt.subplots(1,3, figsize=[20,5], sharex=True, sharey=True)
fig.suptitle('Histogram RGB plot')
fig.text(0.09, 0.5, 'Jumlah Kemunculan', va='center', rotation='vertical')
fig.text(0.5, 0.04, 'Intensitas Warna', ha='center')
axs[0].bar(names, red, color='red')
axs[1].bar(names, green, color='green')
axs[2].bar(names, blue, color='blue')
plt.show
```
